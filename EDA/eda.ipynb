```python
import pandas as pd
import numpy as np
import plotly.express as px
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv("train.csv")
data.head(5)
```

```python
data.dtypes.to_frame()
```

```python
data.isnull().sum().to_frame()
```

```python
data.shape
```

```python
data = data.drop(['Embarked', 'Ticket'], axis=1)

data.head(5)
```

```python
data.nunique().to_frame()
```

```python
mode = data['Cabin'].mode()
mode
```

```python
data = data.fillna(mode[0])
data.head(5)
```

```python
data['Sex'] = data['Sex'].replace({'female': 1, 'male': 0})
data.head(5)
```

```python
fig = px.histogram(data, x=data['Age'])
fig.update_layout(width=1000, height=300)
fig.show()
```

```python
data['Age'] = pd.to_numeric(data['Age'], errors='coerce')

```

```python
data.head(100)
```

```python
median = data['Age'].median()
median
```

```python
data['Age'] = data['Age'].fillna(median)
data.head(100)
```

```python
data.isnull().sum().to_frame()
```

```python
np.argwhere(data['Age'] < 1)
```

```python
data['Age'] = np.where(data['Age'] < 1, 0, data['Age'])
np.argwhere(data['Age'] == 0)
```

```python
data['Age'] = data['Age'].astype(int)
```

```python
for col in data:
    fig = px.histogram(data, x=data[col])
    fig.update_layout(width=1000, height=300)
    fig.show()
```

```python
def find_outliers_IQR(df_col):

   q1=df_col.quantile(0.25)

   q3=df_col.quantile(0.75)

   IQR=q3-q1

   outliers = df_col[((df_col<(q1-1.5*IQR)) | (df_col>(q3+1.5*IQR)))]

   return outliers.index



for col in data:
     if (data[col].dtype==int): 
        outliers = find_outliers_IQR(data[col])
        print(outliers)
        data.drop(index=outliers, inplace=True)

data
```
